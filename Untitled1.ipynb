```python
from ultralytics import YOLO
import cv2

# Load your trained model (replace with the actual path to your weights file)
model = YOLO("runs/detect/train4/weights/best.pt")

# Read an input image
image = cv2.imread("85981e56aea5af4af22a524206.jpg")

# Run prediction
results = model.detect(image, conf=0.5)

# Option 1: Use YOLO's built-in plotting and saving
for i, r in enumerate(results):
    result_img = r.plot()  
    cv2.imwrite(f"result_{i}.jpg", result_img)

# Option 2: Manually draw bounding boxes if you want full control
for result in results:
    for box in result.boxes:
        x1, y1, x2, y2 = map(int, box.xyxy[0])
        cls_id = int(box.cls[0])
        label = result.names[cls_id]
        cv2.rectangle(image, (x1, y1), (x2, y2), (0, 255, 0), 2)
        cv2.putText(image, label, (x1, y1 - 10),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.8, (0, 255, 0), 2)

cv2.imwrite("custom_result1.jpg", image)
```
