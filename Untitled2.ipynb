```python
from djitellopy import Tello
from ultralytics import YOLO
import cv2

# Initialize Tello
tello = Tello()
tello.connect()
print(f"Battery: {tello.get_battery()}%")

# Start video stream
tello.streamon()

# Load your YOLOv11 model (replace with your custom weights if needed)
model = YOLO("runs/detect/train4/weights/best.pt")

# Create OpenCV window
cv2.namedWindow("Tello YOLOv11", cv2.WINDOW_NORMAL)

try:
    while True:
        # Get frame from Tello
        frame = tello.get_frame_read().frame

        # Resize for faster inference (optional)
        frame = cv2.resize(frame, (640, 480))

        # Run YOLO detection
        results = model.predict(frame, conf=0.3, classes=[0, 1])  
        # 👆 set classes=[...] if you only want certain ones, otherwise remove it

        # Plot YOLO results on frame
        annotated_frame = results[0].plot()

        # Show result
        cv2.imshow("Tello YOLOv11", annotated_frame)

        # Quit on 'q' key
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

finally:
    tello.streamoff()
    tello.end()
    cv2.destroyAllWindows()
```

```python
from djitellopy import Tello
from ultralytics import YOLO
import cv2

# Initialize drone
tello = Tello()
tello.connect()
print(f"Battery: {tello.get_battery()}%")

# Start video stream
tello.streamon()

# Load YOLOv11 model (replace with your trained weights if needed)
model = YOLO("runs/detect/train4/weights/best.pt")

# Setup window
cv2.namedWindow("Tello YOLOv11", cv2.WINDOW_NORMAL)

try:
    # Takeoff
    tello.takeoff()

    while True:
        # Get frame from Tello
        frame = tello.get_frame_read().frame
        frame = cv2.resize(frame, (640, 480))

        # Run YOLO detection
        results = model.predict(frame, conf=0.3)

        # Annotate frame
        annotated_frame = results[0].plot()

        # Show in window
        cv2.imshow("Tello YOLOv11", annotated_frame)

        # ---- Keyboard controls ----
        key = cv2.waitKey(1) & 0xFF

        if key == ord('q'):  # quit
            break
        elif key == ord('w'):  # forward
            tello.move_forward(30)
        elif key == ord('s'):  # backward
            tello.move_back(30)
        elif key == ord('a'):  # left
            tello.move_left(30)
        elif key == ord('d'):  # right
            tello.move_right(30)
        elif key == ord('e'):  # rotate clockwise
            tello.rotate_clockwise(30)
        elif key == ord('r'):  # up
            tello.move_up(30)
        elif key == ord('f'):  # down
            tello.move_down(30)
        elif key == ord('l'):  # land
            tello.land()

finally:
    tello.streamoff()
    tello.end()
    cv2.destroyAllWindows()
```
